<h2>Project Overview</h2>

<h5>This repository contains a data analysis solution designed to parse, clean, and extract critical insights from raw Instagram user profile datasets. The project is framed around a fun, gamified scenario where the objective is to process a massive raw dump of follower data under constraint and answer specific audience metrics.</h5>

<h5>The main objective of this project is to showcase string manipulation, data chunking, type casting, and statistical aggregation techniques using Python and Jupyter Notebooks.</h5>

<h5>Scenario: Imagine receiving a dataset containing four Instagram profiles. Your task is to parse, clean, and analyze the data to extract useful insights such as the number of posts, followers, following, and user bios.</h5>

<h6>Questions Addressed:</h6>
<ol>
    <li> Who has the maximum posts?</li>
    <li> Who has the maximum followers?</li>
    <li> Who follows the maximum number of people?</li>
    <li> How many categories are there?</li>


```python
with open("instaprofiles.txt", encoding='utf-8') as f:
    data = f.read()
```

```python
chunks = data.split('\n\n')
chunks = [ c for c in chunks if len(c)>3 ]
# chunks[0]
```

```python
def pars_chunk(chunk):
    chunk=chunk.strip()
    
    sep_chunk=chunk.split('\n')
    username=sep_chunk[0]
    post=(sep_chunk[1].split("postt")[0].replace(",",""))
    followers=(sep_chunk[2].split("followerss")[0].replace(",",""))
    following=(sep_chunk[3].split("followingg")[0])
    category=(sep_chunk[4])
    bio="\n".join(sep_chunk[5:])
    # print(username,post,followers,following,bio,sep="\n")
    return {"UserName":username,"No_Of_Post":post,"No_of_Followers":followers,"No_of_Following":following,"Category":category,"Bio":bio}
# pars_chunk(chunks[0])

all_chunks=[]
for chunk in chunks:
    parsed_chunk= pars_chunk(chunk)
    all_chunks.append(parsed_chunk)
```

```python
import json
s= json.dumps(all_chunks, indent=4)
print(s)
with open("instadetails.json","w")as f:
    f.write(s)
```

<h1>Who has MAX Post?</h1>

```python
max_val = 0
chunk_with_max_post = None

for chunk in all_chunks:
    
    post_str = chunk["No_Of_Post"].split()[0]
    post_count = int(post_str.replace(',', ''))
    
    if max_val < post_count:
        max_val = post_count
        chunk_with_max_post = chunk

print(chunk_with_max_post['UserName'])
```

<h1>Who has MAX Followers?</h1>

```python
max_val = 0
chunk_with_max_followers = None

for chunk in all_chunks:
    
    followers_str = chunk["No_of_Followers"].split()[0]
    followers_count = int(followers_str.replace(',', ''))
    
    
    if max_val < followers_count:
        max_val = followers_count
        chunk_with_max_followers = chunk

print(chunk_with_max_followers['UserName'])
```

<h1>Who has MAX Following?</h1>

```python
max_val = 0
chunk_with_max_followeing = None

for chunk in all_chunks:
    
    following_str = chunk["No_of_Followers"].split()[0]
    following_count = int(following_str.replace(',', ''))
    
    
    if max_val < following_count:
        max_val = following_count
        chunk_with_max_following = chunk

print(chunk_with_max_following['UserName'])
```

<h1>How many categories are there?</h1>

```python
categories = set()
for chunk in all_chunks:
    categories.add(chunk['Category'])
print(f"there are {len(categories)} categories")
    
```
