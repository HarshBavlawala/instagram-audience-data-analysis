```python
with open("instaprofiles.txt", encoding='utf-8') as f:
    data = f.read()
```

```python
chunks = data.split('\n\n')
chunks = [ c for c in chunks if len(c)>3]
```

```python
def pars_chunk(chunk):
    chunk=chunk.strip()
    sep_chunk=chunk.split('\n')
    username=sep_chunk[0]
    post=(sep_chunk[1].split("postt")[0].replace(",",""))
    followers=(sep_chunk[2].split("followerss")[0].replace(",",""))
    following=(sep_chunk[3].split("followingg")[0])
    category=(sep_chunk[4])
    bio="\n".join(sep_chunk[5:])
    # print(username,post,followers,following,bio,sep="\n")
    return {"UserName":username,"No_Of_Post":post,"No_of_Followers":followers,"No_of_Following":following,"Category":category,"Bio":bio}
# pars_chunk(chunks[0])

all_chunks=[]
for chunk in chunks:
    parsed_chunk= pars_chunk(chunk)
    all_chunks.append(parsed_chunk)
```

```python
import json
s= json.dumps(all_chunks, indent=4)
print(s)
with open("instadetails.json","w")as f:
    f.write(s)
```

Who has MAX Post?

```python
max_val = 0
chunk_with_max_post = None

for chunk in all_chunks:
    
    post_str = chunk["No_Of_Post"].split()[0]
    post_count = int(post_str.replace(',', ''))
    
    if max_val < post_count:
        max_val = post_count
        chunk_with_max_post = chunk

print(chunk_with_max_post['UserName'])
```
